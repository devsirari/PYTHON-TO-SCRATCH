```python
# Complete Beginner example

import numpy as np

# Create an array
marks = np.array([10,20,30,40,50])
# Display an array
print(marks)
# Numbers of students
print("no. of students", len(marks))
# Average marks
print("marks:",np.mean(marks))
# Student scoring 60 or more
print("Marks more than 60:", marks[marks >= 30])
```
