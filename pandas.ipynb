```python
import pandas as pd
df = pd.DataFrame({
"Name": ["Rahul","Priya","Amit","Neha"],
"City": ["Delhi","Mumbai","Banglore","Pune"],
"Sales": [5000,7000,4000,6000]
})
print(df)


df[df["Sales"]>5000]
df[df["City"]!= "Delhi"]
print(df[(df["City"] == "Delhi") & (df["Sales"] > 4000)])
print(df[(df["City"] == "Delhi") | (df["City"] == "Mumbai")])
print(df[df["City"].isin(["Delhi", "Pune"])])
print(df.sort_values("Sales"))

```

```python
# Merging DataFrames

import pandas as pd
students = pd.DataFrame({
    "student_id": [101,102,103,104],
    "Name": ["Rahul", "Aman", "Priya", "Raj"]
    })

marks = pd.DataFrame({
    "student_id": [101,102,103,104],
    "Marks": [95,78,90,85]
})

result = pd.merge(students,marks, on="student_id")
```

```python
result
```

```python
# Types of Join

# Inner Join
# Left Join
# Right Join
# Outer Join
```

```python
pd.merge(students, marks, on="student_id", how="inner")
```

```python
students
```

```python
pd.merge(students, marks, on="student_id", how="left")
```

```python
pd.merge(students, marks, on="student_id", how= "right")
```

```python
pd.merge(students, marks, on="student_id", how="outer")
```
