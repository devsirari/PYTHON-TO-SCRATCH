```python
import json

# WRITE
students = [
    {
                "name" : "Rahul",
                "age" : 22,
                "marks" : 91
    },
    { "name" : "priya",
     "age" : 21,
     "grade" : 85
    }
]

with open("students.json", "w") as file:
    json.dump(students, file, indent = 4)

# READ
with open("students.json", "r") as file:
    data = json.load(file)
print(data)

# APPEND/ ADD NEW RECORD
data.append({
    "name" : "Amit",
    "age" : 24,
    "marks" : 99
})

with open("students.json", "w") as file:
    json.dump(data, file, indent =4)

```

```python
# Write the correct data

# Complete CSV example for Students
import csv

studentss = [
    ["Name", "Age", "Marks"],
    ["Rahul", 20, 85],
    ["Priya", 22, 90],
]




with open("studentss.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(studentss)

# Append a new student
with open("studentss.csv", "a", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Amit", 23, 88])

# Read and display the file
with open("studentss.csv", "r", newline="") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
import csv

data = [
    {"Name", "Department", "Salary"},
    {"Rahul", "IT", 50000},
    {"Priya", "HR", 60000},
    {"Amit", "IT", 55000},
    {"Sneha", "IT", 65000}
    ]


with open("employee.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)
```

```python
import csv

total = 0

with open("employees.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        if row["Department"] == "IT":
            total += int(row["Salary"])

print("Total IT salary:", total)
```
