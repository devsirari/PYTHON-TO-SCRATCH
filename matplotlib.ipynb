```python

import matplotlib.pyplot as plt

# 1. Prepare data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# 2. Create a line plot
plt.plot(x, y)

# 3. Add details
plt.title("My First Line Plot")
plt.xlabel("X Axis Label")
plt.ylabel("Y Axis Label")

# 4. Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Prepare data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# 2. Create a line plot
plt.bar(x, y)

# 3. Add details
plt.title("My First Line bar")
plt.xlabel("X Axis Label")
plt.ylabel("Y Axis Label")

# 4. Display the plot
plt.show()

# Create the figure (canvas) and axes (the actual plot area)
fig, ax = plt.subplots()

# Plot data directly onto the axes object
ax.plot([1, 2, 3, 4], [10, 40, 50, 60], color="black", linestyle="--", marker="o")

# Set titles and labels using 'set_' methods
ax.set_title("Customized Square Numbers")
ax.set_xlabel("Inputs")
ax.set_ylabel("Squares")

# Save the figure to your computer instead of just showing it
fig.savefig("my_plot.png")

plt.show()
```

```python
# Pie Chart with explode

expenses = [3000, 2000, 1500, 1000,1500]
explode = [0,0,0,0.1,0]
categories = ["Food", "Rent", "Travel", "Other","dev"]

plt.pie(expenses, labels=categories, explode = explode, autopct="%1.1f%%")

plt.show()
```

```python
plt.pie(
    expenses,
    labels=categories,
    autopct="%1.1f%%"
)
```

```python
months = ["Jan", "Feb", "Mar", "Apr"]

sales_2025 = [100, 150, 130, 180]
sales_2026 = [120, 160, 150, 200]

plt.plot(months, sales_2025, label="2025")
plt.plot(months, sales_2026, label="2026")


plt.legend(loc="upper right")
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr"]

sales_A = [100, 120, 150, 180]
sales_B = [80, 110, 140, 160]

plt.plot(months, sales_A,marker = "o", label="sales A")
plt.plot(months, sales_B,marker = "o", label="sales B")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Product Sales")

plt.legend()
plt.grid()

plt.show()

```

```python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [80, 100, 190, 180]

plt.plot(months, sales)

plt.annotate(
    "Highest Sales",
    xy=("Apr", 180),
    xytext=("Mar", 190),
    arrowprops=dict(arrowstyle="->")
)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
# subplots

import matplotlib.pyplot as plt
# Figure size

plt.figure(figsize=(4,2))
x = [1,2,3,4,5]
y = [10,20,15,25,30]
plt.subplot(1,2,1)
plt.plot(x,y)
plt.title("Line graph")
plt.subplot(1,2,2)
plt.bar(x,y)
plt.title("Bar Chart")
# save a figure
plt.savefig("graph.png")

plt.show()
```

```python
# Changing the X - axis Range and Y - axis Range and # adding text to a graph
x = [1,2,3,4,5]
y = [10,20,15,25,30]
plt.figure(figsize=(4,2))
plt.plot(x,y)
plt.text(5,30, "highest point")
plt.text(1,10, "Lowest Point")
plt.xlim(1,5)
plt.ylim(1,50)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,20,15,25,30]

plt.figure(figsize=(4,2))
plt.plot(x, y, marker="o")
plt.annotate(
    "highest",
    xy=(5,30),
    xytext=(3.5, 32),
arrowprops=dict(arrowstyle="->")
)
plt.show()
```
