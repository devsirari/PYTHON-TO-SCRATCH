```python
# Seaborn
# Seaborn is a Python data visualization library based on matplotlib.

import seaborn as sns
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,20,15,25,30]
plt.figure(figsize=(4,2))
sns.lineplot(x=x, y=y, marker = "o")
plt.title('My First Seaborn Graph')
plt.xlabel("Days")
plt.ylabel("Sales")
plt.show()

```

```python
  # Bar Plot

import seaborn as sns
import matplotlib.pyplot as plt

subjects = ['Math', 'Science', 'English', 'History', 'Art']
marks = [85, 90, 78, 92, 88]

plt.figure(figsize=(4,2))
sns.barplot(x=subjects, y=marks)
plt.title('Student Marks in Different Subjects')
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.show()
```

```python
# Count Plot
# A count plot counnts how many times each category appears

import seaborn as sns
import matplotlib.pyplot as plt

subjects = ["A", "B", "A", "C", "B", "A", "C", "B", "A"]

plt.figure(figsize=(4,2))
sns.countplot(x=subjects)
plt.title("Students Category Count")
plt.show()
```

```python
# Count Plot with a DataFrame

import seaborn as sns
import matplotlib.pyplot as plt



data = {
    "Student": ["John", "Alice", "Bob", "John", "Alice", "Bob", "John", "Alice", "Bob"],
    "subjects": ["Math", "English", "Math", "Science", "Science", "Science", "English", "English", "English"]
}

df = pd.DataFrame(data)
plt.figure(figsize=(4,2))
sns.countplot(data=df, x="Student", hue="subjects")
plt.title("Marksheet")
plt.show()
```

```python
# Box Plot
import seaborn as sns
import matplotlib.pyplot as plt

marks = [85, 90, 78, 92, 88, 95, 80, 75, 82, 89]
plt.figure(figsize=(4,2))
sns.boxplot(data=marks)
plt.title("Distribution of Marks")
plt.show()
print("Mean:", sum(marks) / len(marks))
```

```python
# Box Plot with Categories

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Subjects": ["Math", "Science", "English", "History", "Math", "Math", "Science", "English", "History", "Math"],
    "Marks": [85, 90, 78, 92, 88, 95, 80, 75, 82, 89]
}
df = pd.DataFrame(data)
plt.figure(figsize=(4,5))
sns.boxplot(data=df, x="Subjects", y="Marks")
plt.title("Distribution of Marks by Subject")
plt.show()
```

```python
# Histogram # kde = kernel density estimate, which is a way to estimate the probability density function of a continuous random variable.
import seaborn as sns
import matplotlib.pyplot as plt

marks = [85, 90, 78, 92, 88, 95, 80, 75, 82, 89]
plt.figure(figsize=(4,2))
sns.histplot(marks, bins=5,kde=True)
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
marks = [85, 90, 78, 92, 88, 95, 80, 75, 82, 89]
plt.figure(figsize=(4,2))
sns.kdeplot(marks)
plt.title("KDE Plot of Marks")
plt.show()
```

```python
# Example of Line Plot with a DataFrame

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Jan", "May", "Feb"],
    "Sales": [100, 120, 150, 130, 180, 160],
    "Year": [2026, 2020, 2025, 2021, 2026, 2021]
    }

df = pd.DataFrame(data)
plt.figure(figsize=(4,2))
sns.lineplot(data=df, x="Month", y="Sales", hue="Year", marker = "o")
plt.title("Monthly Sales by Year")
plt.show()
```

```python
# Pair plot
# 

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Math": [80,90,70,85,95],
    "Science": [75,88,72,80,92],
    "English": [85,82,78,90,88],
    "Group": ["A", "B", "A", "B", "A"]
}
df = pd.DataFrame(data)

sns.pairplot(df, hue="Group")
plt.show()
```

```python
# Heatmap
# Heatmaps are a way to visualize data in a matrix format, where the values are represented by colors.
data = [
    [10,20,30],
    [40,50,60],
    [70,80,90]
]
plt.figure(figsize=(4,2))
sns.heatmap(data, annot=True, fmt="d", cmap="YlGnBu")
plt.title("Simple Heatmap")
plt.show()
```

```python
# Correlation 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 45, 55, 70, 85],
    "Sleep_Hours": [8, 7, 7, 6, 6]
})

correlation = data.corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Strip Plot
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Department": ["HR", "HR", "HR", "IT", "IT", "IT"],
    "Salary": [30000, 35000, 32000, 45000, 50000, 48000]
}

sns.stripplot(x="Department", y="Salary", data=data)

plt.title("Salary by Department")
plt.show()
```

```python
# Joint Plot
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7],
    "Marks": [40, 45, 52, 60, 68, 75, 82]
}

sns.jointplot(
    x="Study_Hours",
    y="Marks",
    data=data,
    kind="scatter"
)

plt.show()
```

```python
# Swarm Plot
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Department": ["HR", "HR", "HR", "HR", "IT", "IT", "IT", "IT"],
    "Salary": [30000, 32000, 32000, 35000, 45000, 48000, 48000, 50000]
}

sns.swarmplot(x="Department", y="Salary", data=data)

plt.title("Salary by Department")
plt.show()
```

```python
# Bar plot with hue

import pandas as pd
import matplotlib.pylab as plt
import seaborn as sns

data = {
    "Subject": ["Math", "Science", "English", "History", "Math"],
    "Marks": [85, 90, 78, 92, 88],
    "Class": ["A", "B", "A", "B", "A"]
}

df = pd.DataFrame(data)

sns.barplot(data=df, x="Subject", y="Marks", hue="Class")
plt.title("Marks by Subject and Class")
plt.show()
```

```python
# Line plot with hue
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [100, 120, 150, 130, 180, 160],
    "Year": [2024, 2024, 2024, 2025, 2025, 2025]
}

df = pd.DataFrame(data)
sns.lineplot(data=df, x="Month", y="Sales", hue="Year",marker = "o")
plt.title("Monthly Sales by Year")
plt.show()
```
